## Lab 03b: merge late data safely

Run the five delivery scenarios and inspect their results. The final Silver table should contain **75 trips totaling SAR 1,875.60**. Your observation below is produced by Spark.

```python
from masar.silver import run_incremental_lab
spark = start_spark(WORK, kafka=False)
try:
    result = run_incremental_lab(spark, SOURCE, WORK)
    validate_stage_result('lab03b_silver', result)
    print(json.dumps({'scope': result['scope'], 'checks': result['checks']}, indent=2))
    # Observed learning output
    spark.read.format('delta').load(str(WORK/'mini_lakehouse/silver/trips')).select('trip_id', 'city', 'fare_sar').orderBy('trip_id').show(5, truncate=False)
    # Your turn: write a small query using the correct business key.
    observed_silver = spark.read.format("delta").load(str(WORK / "mini_lakehouse/silver/trips"))
    observed_silver.createOrReplaceTempView("student_silver_trips")
    student_sql = "SELECT COUNT(*) AS trips, SUM(fare_sar) AS total_fare FROM student_silver_trips"
    observed = spark.sql(student_sql).first()
    metrics = {"bronze_trip_receipts": spark.read.format("delta").load(str(WORK / "mini_lakehouse/bronze/trips")).count(),
               "silver_trips": int(observed.trips), "total_fare_sar": format(observed.total_fare, ".2f")}
    print("My SQL result:", observed.asDict())
finally:
    spark.stop()
```
